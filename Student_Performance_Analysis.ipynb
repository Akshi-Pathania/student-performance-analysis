**Dataset Description**

The Academic Survival and Student Retention dataset contains longitudinal information about students' academic performance, financial conditions, attendance, engagement, and student retention.

It tracks student information across multiple semesters, making it possible to analyze how academic performance and other factors change over time.

The dataset includes the following categories of information:
Category	Information
Demographic	Age, gender, household size
Financial	Family income, tuition, scholarship
Academic	Semester GPA, course load, failed courses
Attendance	Attendance percentage
Engagement	LMS logins, advising visits
Student retention	Dropout status and next-semester dropout indicator

**Note on data type:** This is a *synthetic* (computer-generated) dataset designed to mimic a real Student Information System. It intentionally contains "dirty data" such as missing values in `Family_Income` and `LMS_Logins` and inconsistent gender labels, which makes it suitable for practising data cleaning.

# **STEP 4: Data Cleaning and Preprocessing**

```python
#Examine the dataset before cleaning
# Create a copy of the original dataset
# Keep a copy of the original dataset for before-and-after comparison
df_before = df.copy()

# Display original dataset information

print("Original Dataset Shape:", df_before.shape)

print("\nMissing Values:")
display(df_before.isnull().sum().to_frame("Missing_Count"))

print("\nTotal Missing Values:", df_before.isnull().sum().sum())

print("\nDuplicate Records:", df_before.duplicated().sum())
```

1. Handle missing values using appropriate techniques.

```python


# Convert invalid attendance values to missing values
df.loc[~df['Attendance'].between(0, 100), 'Attendance'] = np.nan

# Identify numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Impute numerical columns with their median values
for col in numerical_cols:
    if df[col].notna().any():
        df[col] = df[col].fillna(df[col].median())

# Identify categorical columns
categorical_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()

# Impute categorical columns with their mode
for col in categorical_cols:
    modes = df[col].mode(dropna=True)
    if not modes.empty:
        df[col] = df[col].fillna(modes.iloc[0])

# Display missing values after cleaning
print("Missing values after cleaning:")
display(df.isna().sum().to_frame("Missing_Count"))

print("Total missing values:", df.isna().sum().sum())
```

2. Identify and handle duplicate records.

```python
# Count duplicate records before removal
duplicate_count = df.duplicated().sum()

print("Duplicate records before removal:", duplicate_count)

# Remove exact duplicate rows
before = len(df)
df = df.drop_duplicates().reset_index(drop=True)

print("Removed", before - len(df), "exact duplicate rows.")
print("Dataset shape after removing duplicates:", df.shape)
```

3. Correct inconsistent data types and column names

```python
# Standardize column names
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(r'[^a-z0-9]+', '_', regex=True)
    .str.strip('_')
)

# Standardize inconsistent gender values
df['gender'] = (
    df['gender'].str.strip().str.lower()
    .replace({
        'f': 'Female',
        'female': 'Female',
        'm': 'Male',
        'male': 'Male',
        'other': 'Other',
        'prefer not to say': 'Prefer not to say'
    })
)

# Correct data types
df['student_id'] = df['student_id'].astype('string')

categorical_cols = [
    'gender',
    'housing_status',
    'end_of_semester_status'
]

for col in categorical_cols:
    df[col] = df[col].astype('category')

# Display updated column names and data types
display(df.dtypes.to_frame("Data_Type"))

print("Updated column names:")
print(df.columns.tolist())

print("Gender categories:")
print(df['gender'].value_counts())
```

4. Detect and analyze outliers.

```python
# Select numerical columns for outlier detection
outlier_cols = [
    'family_income',
    'work_hours',
    'emergency_expense',
    'sem_gpa',
    'attendance',
    'lms_logins',
    'financial_stress'
]

outlier_results = []

# Detect outliers using the IQR method
for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_results.append({
        'Column': col,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers)
    })

# Display outlier analysis
outlier_df = pd.DataFrame(outlier_results)
display(outlier_df)
```

5. Compare the dataset before and after cleaning.

```python
# Create a comparison of the original and cleaned datasets
comparison = pd.DataFrame({
    'Before Cleaning': [
        df_before.shape[0],
        df_before.shape[1],
        df_before.isnull().sum().sum(),
        df_before.duplicated().sum(),
        (~df_before['Attendance'].between(0, 100)).sum()
    ],

    'After Cleaning': [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        (~df['attendance'].between(0, 100)).sum()
    ]
}, index=[
    'Number of Rows',
    'Number of Columns',
    'Missing Values',
    'Duplicate Records',
    'Invalid Attendance Values'
])

display(comparison)

# Compare summary statistics before and after cleaning
print("Before Cleaning:")
display(df_before[['Family_Income', 'Sem_GPA', 'Attendance']].describe())

print("After Cleaning:")
display(df[['family_income', 'sem_gpa', 'attendance']].describe())
```

```python
# Group student records by semester
semester_analysis = df.groupby('semester', observed=True).agg(
    Average_GPA=('sem_gpa', 'mean'),
    Average_Attendance=('attendance', 'mean'),
    Average_Work_Hours=('work_hours', 'mean'),
    Average_Financial_Stress=('financial_stress', 'mean'),
    Dropout_Rate=('target_dropout_next_sem', 'mean'),
    Student_Records=('student_id', 'count')
).reset_index()

# Convert dropout rate into percentage
semester_analysis['Dropout_Rate'] = (
    semester_analysis['Dropout_Rate'] * 100
)

display(semester_analysis.round(2))
```
